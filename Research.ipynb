```python
# =========================================
# LIBRARIES
# =========================================
!pip install xgboost shap

import pandas as pd
import numpy as np
np.random.seed(42)
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.pipeline import Pipeline
from sklearn.base import BaseEstimator, TransformerMixin
from scipy.stats.mstats import winsorize
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

import shap
from google.colab import files

import os
import json

os.makedirs("outputs", exist_ok=True)
```

```python
# =========================================
# PIPELINE CUSTOM TRANSFORMER
# =========================================
class Winsorizer(BaseEstimator, TransformerMixin):
    def __init__(self, limits=(0.05, 0.05)):
        self.limits = limits

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = np.array(X, copy=True)
        for i in range(X.shape[1]):
            X[:, i] = winsorize(X[:, i], limits=self.limits)
        return X
```

```python
# =========================================
# LOAD DATA
# =========================================
uploaded = files.upload()
df = pd.read_csv('heart_disease.csv')
```

```python
display(df.head())
print("Dataset shape:", df.shape)
print("Duplicates:", df.duplicated().sum())
print(df["target_binary"].value_counts())
print(df.isnull().sum())
```

```python
# =========================================
# SPLIT
# =========================================
x = df.drop(['target_binary', "num"], axis=1)
y = df['target_binary']

X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
# =========================================
# EVALUATION FUNCTION
# =========================================
def evaluate_model(model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    return {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred)
    }
```

```python
# ===============
# CV FUNCTION
# ===============
def evaluate_with_cv(model, X, y, seeds=[42,52,62], cv=5):
    results = []

    for seed in seeds:
        kfold = StratifiedKFold(n_splits=cv, shuffle=True, random_state=seed)
        scores = cross_val_score(model, X, y, cv=kfold, scoring='f1')
        results.append(scores.mean())

    return np.mean(results), np.std(results)
```

```python
# ===================
# RESULTS STORAGE
# ===================
results = []
cv_results = []
```

```python
# =========================================
# BASELINE PIPELINE
# =========================================
for name, model in {
    "RF": RandomForestClassifier(random_state=42),
    "XGB": XGBClassifier(eval_metric='logloss', random_state=42)
}.items():

    pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", model)
    ])

    res = evaluate_model(pipeline, X_train, X_test, y_train, y_test)
    mean_cv, std_cv = evaluate_with_cv(pipeline, X_train, y_train)

    results.append(["Baseline", name, res["F1"]])
    cv_results.append(["Baseline", name, mean_cv, std_cv])
```

```python
# =========================================
# WINSORIZATION PIPELINE
# =========================================
for name, model in {
    "RF": RandomForestClassifier(random_state=42),
    "XGB": XGBClassifier(eval_metric='logloss', random_state=42)
}.items():

    pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("winsorizer", Winsorizer()),
        ("model", model)
    ])

    res = evaluate_model(pipeline, X_train, X_test, y_train, y_test)
    mean_cv, std_cv = evaluate_with_cv(pipeline, X_train, y_train)

    results.append(["Winsor", name, res["F1"]])
    cv_results.append(["Winsor", name, mean_cv, std_cv])
```

```python
# =========================================
# INTRODUCE MISSING DATA
# =========================================
def introduce_missing(X, rate=0.1, seed=42):
    np.random.seed(seed)
    X_missing = X.copy()
    mask = np.random.rand(*X.shape) < rate
    X_missing[mask] = np.nan
    return X_missing

X_train_miss = introduce_missing(X_train, 0.1)
X_test_miss = introduce_missing(X_test, 0.1)
```

```python
# =========================================
# MEAN IMPUTATION (PIPELINE)
# =========================================
for name, model in {
    "RF": RandomForestClassifier(random_state=42),
    "XGB": XGBClassifier(eval_metric='logloss', random_state=42)
}.items():

    pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="mean")),
        ("model", model)
    ])

    res = evaluate_model(pipeline, X_train_miss, X_test_miss, y_train, y_test)
    mean_cv, std_cv = evaluate_with_cv(pipeline, X_train_miss, y_train)

    results.append(["Mean", name, res["F1"]])
    cv_results.append(["Mean", name, mean_cv, std_cv])
```

```python
# =========================================
#  KNN IMPUTATION (PIPELINE)
# =========================================
for name, model in {
    "RF": RandomForestClassifier(random_state=42),
    "XGB": XGBClassifier(eval_metric='logloss', random_state=42)
}.items():

    pipeline = Pipeline([
        ("imputer", KNNImputer(n_neighbors=5)),
        ("model", model)
    ])

    res = evaluate_model(pipeline, X_train_miss, X_test_miss, y_train, y_test)
    mean_cv, std_cv = evaluate_with_cv(pipeline, X_train_miss, y_train)

    results.append(["KNN", name, res["F1"]])
    cv_results.append(["KNN", name, mean_cv, std_cv])
```

```python
# =========================================
# RESULTS TABLES
# =========================================
results_df = pd.DataFrame(results, columns=["Experiment", "Model", "F1"])
cv_results_df = pd.DataFrame(cv_results, columns=["Experiment", "Model", "Mean_F1", "Std"])

print("\nTest Results:")
display(results_df)

print("\nCross-Validation Results:")
display(cv_results_df)
```

```python
# =========================================
# SAVE RESULTS
# =========================================

results_df.to_csv("outputs/test_results.csv", index=False)
cv_results_df.to_csv("outputs/cv_results.csv", index=False)


with open("outputs/test_results.json", "w") as f:
    json.dump(results, f, indent=4)

with open("outputs/cv_results.json", "w") as f:
    json.dump(cv_results, f, indent=4)
```

```python
# =========================================
# SHAP (PIPELINE)
# =========================================
pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", XGBClassifier(eval_metric='logloss', random_state=42))
])

pipeline.fit(X_train, y_train)

model = pipeline.named_steps["model"]
X_transformed = pipeline[:-1].transform(X_test)

explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_transformed)

feature_names = X_train.columns

shap.summary_plot(
    shap_values,
    X_transformed,
    feature_names=feature_names,
    show=False
)

plt.title("Features Influencing Model Predictions")
plt.tight_layout()

plt.savefig(
    "outputs/shap_summary.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ---------------------------------------------------------
# Create an easy-to-understand feature contribution table
# ---------------------------------------------------------


# Use the first test observation as an example
sample_index = 3

explanation = pd.DataFrame({
    "Feature": feature_names,
    "Patient Value": X_test.iloc[sample_index].values,
    "SHAP Impact": shap_values[sample_index]
})

# Explain the direction of each contribution
explanation["Interpretation"] = np.where(
    explanation["SHAP Impact"] > 0,
    "Increases predicted risk",
    "Decreases predicted risk"
)

# Find the features with the strongest influence
explanation["Impact Strength"] = (
    explanation["SHAP Impact"].abs()
)

top_features = (
    explanation
    .sort_values("Impact Strength", ascending=False)
    .head(5)
)

# ---------------------------------------------------------
# Display the results
# ---------------------------------------------------------

print("\nMODEL EXPLANATION")
print("=" * 60)

# Model prediction
prediction = pipeline.predict(
    X_test.iloc[[sample_index]]
)[0]

probability = pipeline.predict_proba(
    X_test.iloc[[sample_index]]
)[0].max()

print(f"Model prediction: {prediction}")
print(f"Predicted probability: {probability:.1%}")

print("\nMAIN FACTORS INFLUENCING THE PREDICTION")
print("-" * 60)

display(
    top_features[
        ["Feature", "Patient Value", "SHAP Impact", "Interpretation"]
    ].reset_index(drop=True)
)

print("\nSUMMARY")
print("-" * 60)

increasing = top_features[
    top_features["SHAP Impact"] > 0
]["Feature"].tolist()

decreasing = top_features[
    top_features["SHAP Impact"] < 0
]["Feature"].tolist()

print(
    "Main factors increasing the predicted risk:",
    ", ".join(increasing) if increasing else "None"
)

print(
    "Main factors decreasing the predicted risk:",
    ", ".join(decreasing) if decreasing else "None"
)
```
